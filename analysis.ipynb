```python
from google.colab import files
uploaded = files.upload() # This will create a 'Choose File' button

import pandas as pd
import numpy as np

# 1. Load the dataset
# Ensure the file 'final_food_delivery_dataset.csv' is in the same directory
df = pd.read_csv('final_food_delivery_dataset.csv')

# Pre-processing: Convert order_date to datetime objects
df['order_date'] = pd.to_datetime(df['order_date'], dayfirst=True)

# --- Analysis for MCQs ---

# Q1: Which city has the highest total revenue (total_amount) from Gold members?
q1 = df[df['membership'] == 'Gold'].groupby('city')['total_amount'].sum().idxmax()
print(f"Q1 Answer: {q1}")

# Q2: Which cuisine has the highest average order value across all orders?
q2 = df.groupby('cuisine')['total_amount'].mean().idxmax()
print(f"Q2 Answer: {q2}")

# Q3: Distinct users with total order value > ₹1000
user_revenue = df.groupby('user_id')['total_amount'].sum()
q3_count = (user_revenue > 1000).sum()
print(f"Q3 Answer: {q3_count} (> 2000)")

# Q4: Which restaurant rating range generated the highest total revenue?
bins = [3.0, 3.5, 4.0, 4.5, 5.0]
labels = ['3.0 – 3.5', '3.6 – 4.0', '4.1 – 4.5', '4.6 – 5.0']
df['rating_range'] = pd.cut(df['rating'], bins=bins, labels=labels, include_lowest=True)
q4 = df.groupby('rating_range', observed=False)['total_amount'].sum().idxmax()
print(f"Q4 Answer: {q4}")

# Q5: Among Gold members, which city has the highest average order value?
q5 = df[df['membership'] == 'Gold'].groupby('city')['total_amount'].mean().idxmax()
print(f"Q5 Answer: {q5}")

# Q7: Percentage of total orders placed by Gold members
gold_pct = round((df['membership'] == 'Gold').sum() / len(df) * 100)
print(f"Q7 Answer: {gold_pct}%")

# Q9: Highest revenue combination (Gold/Regular + Cuisine)
q9_combos = df.groupby(['membership', 'cuisine'])['total_amount'].sum().idxmax()
print(f"Q9 Answer: {' + '.join(q9_combos)}")

# Q10: Quarter with highest revenue
df['quarter'] = df['order_date'].dt.quarter
q10 = df.groupby('quarter')['total_amount'].sum().idxmax()
print(f"Q10 Answer: Q{q10}")

# --- Additional Stats Requested ---

# Total orders by Gold membership
print(f"Total Gold orders: {len(df[df['membership'] == 'Gold'])}")

# Total revenue from Hyderabad (rounded)
hyd_rev = round(df[df['city'] == 'Hyderabad']['total_amount'].sum())
print(f"Hyderabad Total Revenue: {hyd_rev}")

# Distinct users
print(f"Distinct Users: {df['user_id'].nunique()}")

# Average order value for Gold members
gold_aov = round(df[df['membership'] == 'Gold']['total_amount'].mean(), 2)
print(f"Gold Member AOV: {gold_aov}")

# Orders for restaurants with rating >= 4.5
high_rating = len(df[df['rating'] >= 4.5])
print(f"Rating >= 4.5 Orders: {high_rating}")
```
